# Probe Verification

Load frozen probes and mark each as keep or reject during manual review.

```python
from pathlib import Path
from IPython.display import display
import ipywidgets as widgets

from src.schema import load_probes

probe_path = Path('../probes/probes.json')
probes = load_probes(probe_path) if probe_path.exists() else []
decisions = {}

index = widgets.IntSlider(value=0, min=0, max=max(len(probes) - 1, 0), step=1, description='Probe')
decision = widgets.ToggleButtons(options=['keep', 'reject'], description='Decision')
notes = widgets.Textarea(description='Notes')
out = widgets.Output()

def render(change=None):
    out.clear_output()
    if not probes:
        with out:
            print('No probes found at', probe_path)
        return
    probe = probes[index.value]
    saved = decisions.get(probe.probe_id, {})
    decision.value = saved.get('decision', 'keep')
    notes.value = saved.get('notes', '')
    with out:
        print(probe)

def save(change=None):
    if probes:
        probe = probes[index.value]
        decisions[probe.probe_id] = {'decision': decision.value, 'notes': notes.value}

index.observe(render, names='value')
decision.observe(save, names='value')
notes.observe(save, names='value')
render()
display(widgets.VBox([index, decision, notes, out]))
```
